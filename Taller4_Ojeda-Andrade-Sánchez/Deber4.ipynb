```python
import pandas as pd

# 1. Crear la lista de gastos (Diccionario)
datos_gastos = {
    'Descripción': ['Almuerzo', 'Pasajes', 'Netflix', 'Supermercado', 'Café'],
    'Monto': [5.50, 1.25, 12.00, 45.00, 2.75]
}

df_gastos = pd.DataFrame(datos_gastos)

# 2. Lógica de Validación: Clasificar el gasto
# Si el gasto es mayor a $10, es un "Gasto Alto", si no, es "Gasto Bajo"
df_gastos['Prioridad'] = df_gastos['Monto'].apply(lambda x: 'ALTO' if x > 10 else 'BAJO')

# 3. Operación Matemática: Supongamos que tenemos un presupuesto de $100
presupuesto_inicial = 100
total_gastado = df_gastos['Monto'].sum()
saldo_final = presupuesto_inicial - total_gastado

# 4. Resultados
print("--- MI TABLA DE GASTOS ---")
print(df_gastos)

print(f"\nDinero total gastado: ${total_gastado}")
print(f"Saldo restante en mi cuenta: ${saldo_final}")
```

```python
import pandas as pd

# 1. CREACIÓN DE DATOS (Fácil de entender: un diccionario de Python)
datos_estudiantes = {
    'Nombre': ['Ana', 'Luis', 'María', 'Pedro', 'Elena'],
    'Matemáticas': [8.5, 6.0, 9.5, 4.0, 7.5],
    'Programación': [9.0, 5.5, 10.0, 5.0, 8.0]
}
# Convertimos el diccionario en un DataFrame (la tabla de Pandas)
df = pd.DataFrame(datos_estudiantes)

# 2. PROCESAMIENTO (Lógica de programación)
# Creamos una nueva columna calculando el promedio de las dos materias
df['Promedio'] = (df['Matemáticas'] + df['Programación']) / 2

# 3. VALIDACIÓN (Uso de condiciones)
# Creamos una columna 'Estado' que diga 'Aprobado' si el promedio es >= 7
df['Estado'] = df['Promedio'].apply(lambda x: 'APROBADO' if x >= 7 else 'REPROBADO')

# 4. FILTRADO (Extraer información específica)
# Queremos ver solo a los estudiantes que reprobaron para ayudarlos
estudiantes_en_riesgo = df[df['Estado'] == 'REPROBADO']
# 5. RESULTADOS
print("--- TABLA COMPLETA DE NOTAS ---")
print(df)

print("\n--- ESTUDIANTES QUE NECESITAN REFUERZO ---")
print(estudiantes_en_riesgo[['Nombre', 'Promedio']])

print(f"\nNota promedio del grupo: {df['Promedio'].mean()}")
```
