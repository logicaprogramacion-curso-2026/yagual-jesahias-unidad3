```python
import numpy as np
import matplotlib.pyplot as plt

# Los valores van de 0 a 255 (tipo uint8)
np.random.seed(42)  # Para resultados reproducibles
imagen_rgb = np.random.randint(0, 256, size=(100, 100, 3), dtype=np.uint8)

# Fórmula: 0.299*R + 0.587*G + 0.114*B
pesos_rgb = np.array([0.299, 0.587, 0.114])
imagen_gris = np.dot(imagen_rgb, pesos_rgb)

promedio_brillo = np.mean(imagen_gris)
max_brillo = np.max(imagen_gris)

print(f"Brillo promedio de la imagen: {promedio_brillo:.2f}")
print(f"Brillo máximo: {max_brillo:.2f}")

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(imagen_rgb)
axes[0].set_title("Imagen Original (RGB)")
axes[0].axis('off')

axes[1].imshow(imagen_gris, cmap='gray')
axes[1].set_title("Escala de Grises (NumPy Dot)")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

v0 = 50.0
angulo_deg = 45
g = 9.81

angulo_rad = np.radians(angulo_deg)

t_vuelo = (2 * v0 * np.sin(angulo_rad)) / g

t = np.linspace(0, t_vuelo, num=100)

x = v0 * np.cos(angulo_rad) * t
y = (v0 * np.sin(angulo_rad) * t) - (0.5 * g * (t ** 2))

altura_maxima = np.max(y)
alcance_maximo = np.max(x)

print(f"Alcance máximo: {alcance_maximo:.2f} metros")
print(f"Altura máxima: {altura_maxima:.2f} metros")

plt.figure(figsize=(8, 4))
plt.plot(x, y, label=f"Ángulo: {angulo_deg}° | V0: {v0} m/s", color='crimson', linewidth=2)
plt.title("Simulación de Tiro Parabólico con NumPy")
plt.xlabel("Distancia Horizontal (m)")
plt.ylabel("Altura (m)")
plt.axhline(0, color='black', linestyle='--', linewidth=0.8)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
